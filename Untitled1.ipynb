```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\92335\Downloads\data (1).csv")
```

```python
df = df.drop(columns=["id", "Unnamed: 32"], errors="ignore")
```

```python
df = df.dropna()
```

```python
print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nFirst five rows:")
print(df.head())
```

```python
y = df["diagnosis"].map({"M": 1, "B": 0}).to_numpy()
X = df.drop(columns=["diagnosis"]).to_numpy(dtype=float)
```

```python
np.random.seed(42)

indices = np.random.permutation(len(X))
split_index = int(0.80 * len(X))

train_indices = indices[:split_index]
test_indices = indices[split_index:]
```

```python
X_train = X[train_indices]
X_test = X[test_indices]
y_train = y[train_indices]
y_test = y[test_indices]
```

```python
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)
```

```python
std[std == 0] = 1

X_train = (X_train - mean) / std
X_test = (X_test - mean) / std

print("\nPreprocessing completed.")
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Number of features:", X_train.shape[1])
```

```python
class SingleLayerPerceptron:

    def __init__(self, learning_rate=0.01, epochs=100):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = 0.0
        self.errors = []

    # Step activation function
    def activation(self, value):
        return 1 if value >= 0 else 0

    # Train the perceptron
    def fit(self, X, y):

        number_of_samples, number_of_features = X.shape

        self.weights = np.zeros(number_of_features)
        self.bias = 0.0

        for epoch in range(self.epochs):
            total_errors = 0

            # Shuffle training samples each epoch
            order = np.random.permutation(number_of_samples)

            for i in order:
                # Weighted sum plus bias
                z = np.dot(X[i], self.weights) + self.bias

                # Predicted class
                prediction = self.activation(z)

                # Calculate error
                error = y[i] - prediction

                # Update weights and bias
                self.weights += (
                    self.learning_rate * error * X[i]
                )
                self.bias += self.learning_rate * error

                if error != 0:
                    total_errors += 1

            self.errors.append(total_errors)

            if (epoch + 1) % 10 == 0:
                print(
                    f"Epoch {epoch + 1}/{self.epochs}, "
                    f"Errors: {total_errors}"
                )

    # Predict class labels
    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        return (linear_output >= 0).astype(int)

```

```python
model = SingleLayerPerceptron(
    learning_rate=0.01,
    epochs=100
)

print("\nTraining started...")
model.fit(X_train, y_train)
print("Training completed!")
```

```python
y_pred = model.predict(X_test)
```

```python
TP = np.sum((y_test == 1) & (y_pred == 1))
TN = np.sum((y_test == 0) & (y_pred == 0))
FP = np.sum((y_test == 0) & (y_pred == 1))
FN = np.sum((y_test == 1) & (y_pred == 0))

accuracy = (TP + TN) / len(y_test)

precision = TP / (TP + FP) if (TP + FP) else 0
recall = TP / (TP + FN) if (TP + FN) else 0

f1_score = (
    2 * precision * recall / (precision + recall)
    if (precision + recall) else 0
)

print("\n========== FINAL RESULTS ==========")
print("True Positives (TP):", TP)
print("True Negatives (TN):", TN)
print("False Positives (FP):", FP)
print("False Negatives (FN):", FN)

print(f"\nAccuracy:  {accuracy * 100:.2f}%")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1_score:.4f}")

print("\nConfusion Matrix:")
print(np.array([[TN, FP],
                [FN, TP]]))
```

```python
plt.figure(figsize=(8, 5))
plt.plot(
    range(1, model.epochs + 1),
    model.errors,
    marker="o",
    markersize=3
)

plt.title("Single-Layer Perceptron Training Errors")
plt.xlabel("Epoch")
plt.ylabel("Number of Misclassified Training Samples")
plt.grid(True)
plt.tight_layout()
plt.show()
```
